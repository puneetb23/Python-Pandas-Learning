```python
# FINDING MISSING DATA
import numpy as np
import pandas as pd
```

```python
data = {
    'A' : [1,2,np.nan, 4,5],   # this nan is the null or missing data
    'B' : [1, 2,3,4,5],
    'C' : [1,2,3, np.nan , np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
# REMOVING MISSSING DATA
df
```

```python
df.dropna() #it removed all the null value
```

```python
df
```

```python
df.dropna(thresh=4)  # threshold keeps only whose have that number of non null values
```

```python
#FILLING THE MISSING DATA
df
```

```python
df.fillna(0)  # fillna fills the value at the position of null  
#but it will not remove or add permanently untill we use , inplace = true
```

```python
values = {'A':0 , 'B' : 100,'c':300,'D':400}
df.fillna(value = values)  # it will fill the values column wise
```

```python
df
```

```python
df.fillna(df.mean())
```
