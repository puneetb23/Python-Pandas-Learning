```python
df["Episodes"] = df["Episodes"].astype(str).str.extract(r'(\d+)')
```

```python
def extract_episodes(txt):
    check = True
    data = ""
    for i in txt:
        if i == 'Episodes':
            check = True
        if check == True:
            data += i
            break
    return data
        
```
